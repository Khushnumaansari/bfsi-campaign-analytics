```python
import pandas as pd
import matplotlib.pyplot as plt

customer = pd.read_csv("dim_customer.csv")
campaign = pd.read_csv("dim_campaign.csv")
experiment = pd.read_csv("fact_experiment.csv")
cost = pd.read_csv("fact_campaign_cost.csv")
```

```python
exp = experiment.merge(campaign, on="campaign_id", how="left")
exp = exp.merge(customer, on="customer_id", how="left")
```

```python
overall = exp["converted"].mean()
print(f"Overall conversion rate: {overall:.2%}")
```

```python
version_perf = (
    exp.groupby("campaign_version")
    .agg(
        customers=("customer_id", "count"),
        conversions=("converted", "sum"),
        conversion_rate=("converted", "mean")
    )
    .reset_index()
)

display(version_perf)
```

```python
channel_perf = (
    exp.groupby("channel")
    .agg(
        customers=("customer_id", "count"),
        conversions=("converted", "sum"),
        conversion_rate=("converted", "mean")
    )
    .reset_index()
)

display(channel_perf)
```

```python
age_perf = (
    exp.groupby("age_group", observed=True)
    .agg(
        customers=("customer_id", "count"),
        conversions=("converted", "sum"),
        conversion_rate=("converted", "mean")
    )
    .reset_index()
)

display(age_perf)
```

```python
budget_perf = (
    cost.groupby(["channel", "campaign_version"])
    .agg(
        customers_contacted=("customers_contacted", "sum"),
        conversions=("conversions", "sum"),
        planned_budget=("planned_budget", "sum"),
        actual_spend=("actual_spend", "sum"),
        estimated_revenue=("estimated_contribution", "sum"),
        estimated_profit=("estimated_profit", "sum")
    )
    .reset_index()
)

display(budget_perf)
```
